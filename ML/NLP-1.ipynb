# Topics
- Corpus - a paragraph
- Documents - sentences
- Vocabulary - unique words present in that paragraph
- Words - all words present in Corpus

# Tokenization 
- Important step in NLP. 
- Tokens obtained from Corpus (paragraph) are called Documents (sentences)
    Example: ```My name is ranga. My friend's name is Ananth. Ananth and I like to play sports.``` When this paragraph tokenized, we have 3 tokens (3 documents/sentences)
    - My name is Ranga
    - My friend's name is Ananth.
    - Ananth and I like to play sports.
- Tokens obtained from Documents (sentences) into Vocabulary (unique words) and Words. In the above example, total number of unique words is 11
    1. My
    2. name
    3. is
    4. Ranga
    5. friend's
    6. Ananth
    7. and
    8. I 
    9. to
    10. play
    11. sports


````python
%pip install nltk
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
````

````python
corpus = """This is the first document.
This document is the second document!
And this is the third one.
Is this the first document?
This docuemnt's purpose is to test sentence tokenization.
"""

print(corpus)
````

````python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(corpus)
print(sentences)
````

## Convert paragraph and sentences into words
- NOTE the tokenization of 's and 

````python
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
````

````python
for s in sentences: 
    print(word_tokenize(s))
````
