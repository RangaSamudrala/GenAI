```python
# Basic class definition with constructor and accessing public members
import os

class Car:
    def __init__(self, manufacturer, model, engine, seats, doors):
        self.manufacturer = manufacturer
        self.model = model
        self.engine = engine
        self.seats = seats
        self.doors = doors

    def describe(self):
        print(f'Manufacturer: {self.manufacturer}')
        print(f'Model: {self.model}')
        print(f'Engine: {self.engine}')
        print(f'Seats: {self.seats}')
        print(f'Doors: {self.doors}')

camry_se = Car('Toyota', 'Camry SE', 'Gas', 5, 2)
camry_se.describe()


```

```python
# Basic class inheritence with constructor and accessing parent constructor
class Toyota(Car):
    def __init__(self, model, engine, seats, doors):
        super().__init__('Toyota', model, engine, seats, doors)

class Tesla(Car):
    def __init__(self, model, seats, doors, is_self_driving):
        super().__init__('Telsa', model, 'Electric', seats, doors)
        self.is_self_driving = True

    def describe(self):
        super().describe()
        print(f'Is Self Driving: {self.is_self_driving}')

t1 = Toyota('Camry SE', 'Gas', 5, 2)
t2 = Tesla('M3', 5, 4, False)

t1.describe()
t2.describe()
```

```python
# Polymorphism extending from more than one parent class
class Animal:
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        print('needs to be implemented by the child class')

class Pet:
    def __init__(self, owner):
        self.owner = owner

class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)
    
    def make_sound(self):
        return f'{self.name} woof, woof'


d1 = Dog('Raja', 'Samudrala')

print(f'{d1.make_sound()}')
```

```python
# Abstract class definition and extension. Private, protected and public members
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    
class Rectangle(Shape):
    def __init__(self, length, width):
        self.__length = length ## private variable
        self.__width = width ## private variable
        self._type = 'Rectangle'  ## protected variable

    def area(self):
        return self.__length * self.__width

    def get_length(self):
        return self.__length
    
    def get_width(self):
        return self.__width

class Circle(Shape):
    def __init__(self, radius):
        self.__radius = radius  ## private variable
        self._type = 'Circle'  ## protected variable
    
    def area(self):
        return math.pi * (self.__radius ** 2)

    def get_radius(self):
        return self.__radius

r1 = Rectangle(4, 5)
c1 = Circle(5)

print(f'Area of rectangle is {r1.area()}')
print(f'Area of circle is {c1.area()}')
print(f'r1 type is {r1._type}')
print(f"r1's length: {r1.get_length()}")
print(f"r1's width: {r1.get_width()}")
print(f"c1's radius: {c1.get_radius()}")
```

```python
# Magic methods
class Computer(ABC):
    
    def __init__(self, os_type):
        self.__os_type__ = os_type

    @abstractmethod
    def print_specifications():
        pass

class Apple(Computer):

    def __init__(self):
        super().__init__('Darwin')

    def print_specifications(self):
        print(f'os_type: {self.__os_type__}')

a = Apple()
a.print_specifications()
```
