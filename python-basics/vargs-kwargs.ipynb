```python
'''
Multiple args
'''
def f1(*args):
	print(args)
```

```python
f1(1, 2, 4)
```

```python
def msum(*args):
	return sum(args)
```

```python
msum(1, 20, 3, 4)
```

```python
def vsum(**kwargs):
	print(kwargs)
	print(kwargs['a'])
```

```python
vsum(a=1, b=2, c=3, d=4)
```

```python
def f2(*args, **kwargs):
	print(f'args: {args}')
	print(f'kwargs: {kwargs}')
```

```python
f2(1, 2, 3, 4, a=10, b=20)
```

```python
f2(1, 2, 3, 4)
```

```python
def retEvens(*args):
    return [x for x in args if x%2 == 0]

retEvens(1, 2, 3, 4, 5, 56, 7, 8, 10)
```

```python
def myfunc(str):
	counter = 0
	returnStr = ''
	for c in str:
		if counter%2 == 0:
			returnStr += c.upper()
		else:
			returnStr += c.lower()
		counter += 1
	return returnStr

print(myfunc('jkdhfjkdsf'))
```
