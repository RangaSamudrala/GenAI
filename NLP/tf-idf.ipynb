# TF-IDF 
Term Frequency - Inverse Document Formula

Term Frequency = # of repetitions of words in sentence / # of words in sentence
Inverse Document Frequency = LogE(# of sentences / # of sentenaces containing the word)

S1 -> good	boy		

S2 -> good 	girl

S3 -> boy		girl	good

Term Frequency													Words 			IDF
				S1			S2		S3	
good		1/2			1/2		1/3								good				logE(3/3) = 0

boy			1/2			0			1/3								boy					logE(3/2)

girl		0				1/2		1/3								good				logE(3/2)


### Final TF*IDF
				good			boy							girl
S1			0				1/2 * LogE(3/2)		0														

S2			0 				0 							1/2 * LogE(3/2)		===> 			

S3			0 				1/3 * LogE(3/2)	1/3 * LogE(3/2)


## Advantages
- Intuitive
- Fixed size
- Word importance is captured.

## Disadvantage
- Still sparse matrix
- Out of vocabulary (OOV)


```python
import pandas as pd
messages=pd.read_csv('spam.csv', sep=',')
# , names=["label","message", 'c1', 'c2', 'c3']
```

```python
messages.head
```

```python
## Data Cleaning And Preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
## Create the Bag OF Words model
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
cv=CountVectorizer(max_features=100,binary=True)
```

```python
X=cv.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, 
    formatter=dict(float=lambda x: "%.3g" % x))
X
```
