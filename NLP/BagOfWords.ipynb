```python
# Data cleaning and preprocessing
import re
import string
from nltk.stem import PorterStemmer, WordNetLemmatizer
ps = PorterStemmer()
wordnet=WordNetLemmatizer()
```

```python
corpus = []
def invoke_stopwords(doc):
    return [w for w in nltk.word_tokenize(doc) if w not in set(stopwords.words('english'))]

def invoke_stopwords_and_stem(doc):
    return [ps.stem(w) for w in nltk.word_tokenize(doc) if w not in set(stopwords.words('english'))]


def remove_special_characters(doc):
    # Remove all the special characters
    review = re.sub(r'\W', ' ', str(doc))
    # remove all single characters
    review = re.sub(r'\s+[a-zA-Z]\s+', ' ', review)
    # Remove single characters from the start
    review = re.sub(r'\^[a-zA-Z]\s+', ' ', review) 
    # Substituting multiple spaces with single space
    review = re.sub(r'\s+', ' ', review, flags=re.I)
    # Removing prefixed 'b'
    review = re.sub(r'^b\s+', '', review)
    # Converting to Lowercase
    review = review.lower()
    return review


```
