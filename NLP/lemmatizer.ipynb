# Lemmatization
- A better way get root words compared to **Stemming**. Output is called **Lemma** which is the root word.
- NLTK provides class **WordNetLemmatizer**. The class provides a function named **morphy()** to the **WordNetCorpusReader** class. 
- Very useful for Q&A, chatbot, text summarization.

```python
# Lemma is a root word

!pip install nltk
import nltk
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer


words = ['ate', 'eat',  'eating', 'eaten', 'finalize', 'finally', 'going', 'gone', 'history', 'histology', 'programming', 'programs', 'writing', 'writes', 'written' ]

lemmatizer = WordNetLemmatizer()

# POSTAGs: Used to provide hints when lemmatizing a word.
# Noun - n
# Verb - v
# Adjective - a
# Adverb - r

hints = ['a', 'n', 'v', 'r']
for word in words:
	for hint in hints:
		print(f"{word}'s stem with hint: {hint} is: {lemmatizer.lemmatize(word, pos=hint)}")

```

```python
lemmatizer.lemmatize("sportingly", pos='n')
```
