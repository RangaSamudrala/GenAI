# Stemming
Process of affixing words to its root word

```python
!pip install nltk
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words = ['ate', 'eat',  'eating', 'eaten', 'finalize', 'finally', 'history', 'histology', 'programming', 'programs', 'writing', 'writes', 'written' ]
for word in words:
	print(f"{word}'s stem is: {stemmer.stem(word)}")
```

```python
# disadvantage
print(f' for congralations, stem word is {stemmer.stem("congratulations")}')
```

```python
from nltk.stem import RegexpStemmer
regexStemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
for word in words:
	print(f"{word}'s stem is: {regexStemmer.stem(word)}")
```

```python
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english')
```

```python
for word in words:
	 print(f'{word} -> {snowball.stem(word)}')
```
